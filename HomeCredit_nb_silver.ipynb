```python
from pyspark.sql import functions as F
from pyspark.sql import Row
from datetime import datetime
import time
import json


start_time = time.time()

batch_id = datetime.utcnow().strftime("%Y%m%d_%H%M%S")

catalog = "insurance"
schema = "homecredit"

source_layer = "BRONZE"

validation_results = []
execution_logs = []

print("Silver Layer Started")
print("Batch ID:", batch_id)
print("Start Time:", datetime.utcnow())
```

```python
from pyspark.sql import functions as F


def add_audit_columns(df, status):


    df = (
        df
        .withColumn(
            "batch_id",
            F.lit(batch_id)
        )
        .withColumn(
            "source_layer",
            F.lit("BRONZE")
        )
        .withColumn(
            "processing_timestamp",
            F.current_timestamp()
        )
        .withColumn(
            "record_status",
            F.lit(status)
        )
    )


    return df
```

```python
def remove_duplicates(df, keys):

    before = df.count()

    clean_df = df.dropDuplicates(keys)

    after = clean_df.count()

    removed = before - after

    return clean_df, before, removed, after



tables_keys = {
    "application_train": ["SK_ID_CURR"],
    "application_test": ["SK_ID_CURR"],
    "bureau": ["SK_ID_BUREAU"],
    "bureau_balance": ["SK_ID_BUREAU"],
    "previous_application": ["SK_ID_PREV"]
}


duplicate_results = []


for table_name, keys in tables_keys.items():

    test_df, before, duplicates, after = remove_duplicates(
        dfs[table_name],
        keys
    )


    duplicate_results.append(
        (
            table_name,
            before,
            duplicates,
            after
        )
    )


    print("=" * 60)
    print("Table Name:", table_name)
    print("Records before:", before)
    print("Duplicates removed:", duplicates)
    print("Records after:", after)



# Create output table

duplicate_summary_df = spark.createDataFrame(
    duplicate_results,
    [
        "table_name",
        "records_before",
        "duplicates_removed",
        "records_after"
    ]
)


display(duplicate_summary_df)
```

```python
from pyspark.sql import functions as F


def handle_nulls(df):

    # Null count before cleaning
    before_nulls = (
        df.select(
            [
                F.count(
                    F.when(
                        F.col(c).isNull(),
                        c
                    )
                ).alias(c)
                for c in df.columns
            ]
        )
    )


    # Handle null values

    for c, dtype in df.dtypes:


        if dtype in [
            "int",
            "bigint",
            "double",
            "float"
        ]:

            df = df.fillna(
                {c: 0}
            )


        else:

            df = df.fillna(
                {c: "UNKNOWN"}
            )



    # Null count after cleaning

    after_nulls = (
        df.select(
            [
                F.count(
                    F.when(
                        F.col(c).isNull(),
                        c
                    )
                ).alias(c)
                for c in df.columns
            ]
        )
    )


    return df, before_nulls, after_nulls



# Tables to process

tables = [
    "application_train",
    "application_test",
    "bureau",
    "bureau_balance",
    "previous_application"
]



# Process all tables

for table_name in tables:


    clean_df, null_before, null_after = handle_nulls(
        dfs[table_name]
    )


    # Save cleaned dataframe

    dfs[table_name + "_clean"] = clean_df



    print("="*60)

    print("Table Name:", table_name)

    print("Records Before:",
          dfs[table_name].count())


    print("Records After:",
          clean_df.count())


    print("Null Count Before Cleaning")

    display(null_before)


    print("Null Count After Cleaning")

    display(null_after)
```

```python
from pyspark.sql import functions as F


# Standardize text function

def standardize_text(df):

    for c, dtype in df.dtypes:

        if dtype == "string":

            df = df.withColumn(
                c,
                F.upper(
                    F.trim(
                        F.col(c)
                    )
                )
            )

    return df



# Tables after null handling

clean_tables = [
    "application_train",
    "application_test",
    "bureau",
    "bureau_balance",
    "previous_application"
]


# Apply standardization for all 5 tables

for table_name in clean_tables:


    print("=" * 60)

    print("Processing Table:", table_name)


    # Take null cleaned dataframe

    clean_df = dfs[table_name + "_clean"]


    # Apply text standardization

    standardized_df = standardize_text(
        clean_df
    )


    # Store output

    dfs[table_name + "_silver"] = standardized_df



    print(
        "Records:",
        standardized_df.count()
    )


    display(
        standardized_df.limit(5)
    )
```

```python
from pyspark.sql import functions as F


def correct_datatypes(df):

    changed = []

    for c in df.columns:

        if c.startswith("AMT_"):

            df = df.withColumn(
                c,
                F.col(c).cast("double")
            )

            changed.append(c)


    return df, changed



tables = [
    "application_train",
    "application_test",
    "bureau",
    "bureau_balance",
    "previous_application"
]


for table_name in tables:


    print("=" * 60)

    print("Table Name:", table_name)


    # take standardized dataframe

    df = dfs[table_name + "_silver"]


    # correct datatypes

    corrected_df, converted_columns = correct_datatypes(df)


    # store result

    dfs[table_name + "_datatype_clean"] = corrected_df



    print(
        "Converted columns:"
    )

    print(
        converted_columns
    )


    print(
        "Records:",
        corrected_df.count()
    )


    display(
        corrected_df.limit(5)
    )
```

```python
from pyspark.sql import functions as F


def apply_business_rules(df, table):


    # Default empty reject dataframe

    rejected = spark.createDataFrame(
        [],
        df.schema
    )


    valid = df



    # Rule 1: Application tables
    if table in [
        "application_train",
        "application_test"
    ]:


        rejected = df.filter(
            F.col("AMT_CREDIT") <= 0
        )


        valid = df.filter(
            F.col("AMT_CREDIT") > 0
        )



    # Rule 2: Bureau credit amount cannot be negative

    elif table == "bureau":


        rejected = df.filter(
            F.col("AMT_CREDIT_SUM") < 0
        )


        valid = df.filter(
            F.col("AMT_CREDIT_SUM") >= 0
        )



    # Rule 3: Bureau balance STATUS validation

    elif table == "bureau_balance":


        allowed_status = [
            "0",
            "1",
            "2",
            "3",
            "4",
            "5",
            "C",
            "X"
        ]


        rejected = df.filter(
            ~F.col("STATUS").isin(allowed_status)
        )


        valid = df.filter(
            F.col("STATUS").isin(allowed_status)
        )



    # Rule 4: Previous application amount validation

    elif table == "previous_application":


        rejected = df.filter(
            F.col("AMT_APPLICATION") < 0
        )


        valid = df.filter(
            F.col("AMT_APPLICATION") >= 0
        )



    print("="*60)

    print("Table Name:", table)

    print(
        "Total Records:",
        df.count()
    )


    print(
        "Valid Records:",
        valid.count()
    )


    print(
        "Rejected Records:",
        rejected.count()
    )


    return valid, rejected
```

```python
tables = [
    "application_train",
    "application_test",
    "bureau",
    "bureau_balance",
    "previous_application"
]


for table_name in tables:


    df = dfs[table_name + "_datatype_clean"]


    valid_df, reject_df = apply_business_rules(
        df,
        table_name
    )


    # Save outputs

    dfs[table_name + "_valid"] = valid_df

    dfs[table_name + "_reject"] = reject_df


    print("Valid dataframe stored:")
    print(table_name + "_valid")


    print("Reject dataframe stored:")
    print(table_name + "_reject")
```

```python
# Tables to write into Silver

tables = [
    "application_train",
    "application_test",
    "bureau",
    "bureau_balance",
    "previous_application"
]


for table_name in tables:


    print("=" * 60)

    print("Creating Silver Table:", table_name)



    # Get valid dataframe from business rules

    valid_df = dfs[table_name + "_valid"]



    # Add audit columns

    silver_df = add_audit_columns(
        valid_df,
        "PROCESSED"
    )



    # Silver table path

    silver_table_name = (
        f"insurance.homecredit.{table_name}_silver"
    )



    # Write Delta table

    silver_df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable(
            silver_table_name
        )



    print(
        "Silver table created:",
        silver_table_name
    )


    print(
        "Records written:",
        silver_df.count()
    )



    display(
        spark.table(
            silver_table_name
        ).limit(10)
    )
```

```python
from pyspark.sql import Row
from pyspark.sql import functions as F
from datetime import datetime


validation_results = []


def add_validation_result(
    table_name,
    rule_name,
    total_records,
    failed_records
):

    validation_results.append(

        Row(

            batch_id=batch_id,

            table_name=table_name,

            validation_rule=rule_name,

            total_records=total_records,

            failed_records=failed_records,

            passed=(failed_records == 0),

            validation_timestamp=datetime.utcnow()

        )
    )



# Tables to validate

tables = [
    "application_train",
    "application_test",
    "bureau",
    "bureau_balance",
    "previous_application"
]



for table_name in tables:


    print("="*60)

    print("Validating Table:", table_name)



    # take silver dataframe

    df = dfs[table_name + "_datatype_clean"]



    total_records = df.count()



    # 1. Null validation

    null_failed = (

        df.filter(

            F.array_contains(

                F.array(

                    [

                    F.col(c).isNull()

                    for c in df.columns

                    ]

                ),

                True

            )

        )

        .count()

    )



    add_validation_result(

        table_name,

        "Null value check",

        total_records,

        null_failed

    )




    # 2. Duplicate validation

    duplicate_count = (

        total_records -

        df.dropDuplicates().count()

    )



    add_validation_result(

        table_name,

        "Duplicate record check",

        total_records,

        duplicate_count

    )




    # 3. Record count validation

    failed_count = 0 if total_records > 0 else 1


    add_validation_result(

        table_name,

        "Record count greater than zero",

        total_records,

        failed_count

    )




    print(
        "Total Records:",
        total_records
    )

    print(
        "Validation Completed"
    )



# Create validation dataframe

validation_df = spark.createDataFrame(
    validation_results
)



display(validation_df)
```

```python
validation_df.write \
.format("delta") \
.mode("overwrite") \
.option("overwriteSchema","true") \
.saveAsTable(
    "insurance.homecredit.silver_validation_report"
)


print("Validation Report Created")


display(
    spark.table(
        "insurance.homecredit.silver_validation_report"
    )
)

print("Validation Report Saved")
```

```python
from pyspark.sql import Row
from datetime import datetime
import time


execution_logs = []


tables = [
    "application_train",
    "application_test",
    "bureau",
    "bureau_balance",
    "previous_application"
]


for table_name in tables:


    start_time = time.time()


    # Processed dataframe

    processed_df = dfs[table_name + "_valid"]


    records_processed = processed_df.count()


    # Rejected dataframe

    rejected_df = dfs[table_name + "_reject"]


    records_rejected = rejected_df.count()



    execution_logs.append(

        Row(

            batch_id=batch_id,

            table_name=table_name,

            source_layer="BRONZE",

            records_received=
                dfs[table_name].count(),

            records_processed=
                records_processed,

            records_rejected=
                records_rejected,

            processing_duration_seconds=
                round(time.time()-start_time,2),

            processing_timestamp=
                datetime.utcnow(),

            record_status="COMPLETED"

        )
    )


    print("="*60)

    print("Table Name:", table_name)

    print("Received:", dfs[table_name].count())

    print("Processed:", records_processed)

    print("Rejected:", records_rejected)



# Create audit dataframe

audit_df = spark.createDataFrame(
    execution_logs
)



display(audit_df)
```

```python
audit_df.write \
.format("delta") \
.mode("overwrite") \
.option("overwriteSchema","true") \
.saveAsTable(
    "insurance.homecredit.silver_audit_report"
)


print("Audit Report Created")


display(
    spark.table(
        "insurance.homecredit.silver_audit_report"
    )
)
```
