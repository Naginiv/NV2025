```python
from pyspark.sql import functions as F
from pyspark.sql import Row
from datetime import datetime
import time


gold_start_time = time.time()

gold_batch_id = datetime.utcnow().strftime("%Y%m%d_%H%M%S")


catalog = "insurance"
schema = "homecredit"


print("Gold Layer Started")
print("Batch ID:", gold_batch_id)
```

```python
tables = [
    "application_train",
    "application_test",
    "bureau",
    "bureau_balance",
    "previous_application"
]


gold_dfs = {}


for table in tables:


    gold_dfs[table] = spark.table(
        f"{catalog}.{schema}.{table}_silver"
    )


    print("="*60)

    print("Loaded Silver Table:", table)

    print(
        "Records:",
        gold_dfs[table].count()
    )


    display(
        gold_dfs[table].limit(5)
    )
```

```python
application_kpi = (

gold_dfs["application_train"]

.groupBy("CODE_GENDER")

.agg(

    F.count("*").alias("total_customers"),

    F.avg("AMT_INCOME_TOTAL")
    .alias("avg_income"),

    F.avg("AMT_CREDIT")
    .alias("avg_credit"),

    F.sum("TARGET")
    .alias("default_count")

)

)


display(application_kpi)
```

```python
application_kpi.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
"insurance.homecredit.gold_customer_kpi"
)


print("KPI Table Created")
```

```python
bureau_summary = (

gold_dfs["bureau"]

.groupBy("CREDIT_ACTIVE")

.agg(

F.count("*")
.alias("total_loans"),

F.sum("AMT_CREDIT_SUM")
.alias("total_credit"),

F.avg("AMT_CREDIT_SUM")
.alias("average_credit")

)

)


display(bureau_summary)
```

```python
bureau_summary.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
"insurance.homecredit.gold_bureau_summary"
)
```

```python
previous_summary = (

gold_dfs["previous_application"]

.groupBy("NAME_CONTRACT_STATUS")

.agg(

F.count("*")
.alias("application_count"),

F.avg("AMT_APPLICATION")
.alias("avg_application_amount"),

F.avg("AMT_CREDIT")
.alias("avg_credit_amount")

)

)


display(previous_summary)
```

```python
previous_summary.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
"insurance.homecredit.gold_previous_application_summary"
)
```

```python
trend_df = (

gold_dfs["application_train"]

.groupBy("DAYS_BIRTH")

.agg(

F.count("*")
.alias("customer_count"),

F.avg("AMT_CREDIT")
.alias("avg_credit")

)

)


display(trend_df.limit(20))
```

```python
trend_df.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
"insurance.homecredit.gold_trend_analysis"
)
```

```python
performance_df = (

gold_dfs["application_train"]

.groupBy("TARGET")

.agg(

F.count("*")
.alias("records"),

F.avg("AMT_INCOME_TOTAL")
.alias("average_income"),

F.avg("AMT_CREDIT")
.alias("average_credit")

)

)


display(performance_df)
```

```python
performance_df.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
"insurance.homecredit.gold_performance_metrics"
)
```

```python
gold_validation=[]


for table in tables:


    source_count = spark.table(
        f"insurance.homecredit.{table}_silver"
    ).count()


    gold_validation.append(

        Row(

            batch_id=gold_batch_id,

            table_name=table,

            source_records=source_count,

            validation_rule="record_count_check",

            passed=True,

            validation_timestamp=datetime.utcnow()

        )

    )



gold_validation_df = spark.createDataFrame(
    gold_validation
)


display(gold_validation_df)
```

```python
gold_validation_df.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
"insurance.homecredit.gold_validation_report"
)
```

```python
gold_audit=[]


for table in tables:


    gold_audit.append(

        Row(

            batch_id=gold_batch_id,

            table_name=table,

            aggregation_date=datetime.utcnow().date(),

            processing_timestamp=datetime.utcnow(),

            gold_load_status="COMPLETED"

        )

    )



audit_df=spark.createDataFrame(
    gold_audit
)


display(audit_df)
```

```python
audit_df.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
"insurance.homecredit.gold_audit_report"
)
```

```python
execution_log = [

Row(

batch_id=gold_batch_id,

source_records=sum(
    [
    gold_dfs[t].count()
    for t in tables
    ]
),

target_records=(

application_kpi.count()
+
bureau_summary.count()
+
previous_summary.count()

),

processing_duration_seconds=
round(
time.time()-gold_start_time,
2
),

kpi_generation_status="SUCCESS",

execution_timestamp=datetime.utcnow()

)

]


execution_df=spark.createDataFrame(
execution_log
)


display(execution_df)
```

```python
execution_df.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
"insurance.homecredit.gold_execution_logs"
)
```
