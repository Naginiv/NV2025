```python
############################ silver layer #########################
# split address line into separate values
from pyspark.sql.functions import split, col
if df_address_silver is not None:
    df_address_silver = df_address_silver.withColumn(
        'address_line_1_parts',
        split(col('address_line_1'), ' ')
    )
    display(df_address_silver)
```

```python
############################ silver layer #########################
# extract year, month and date from order date
from pyspark.sql.functions import year, month, dayofmonth

if df_sales_order_header_silver is not None:
    df_sales_order_header_silver = df_sales_order_header_silver.withColumn(
        'order_year', year(col('order_date'))
    ).withColumn(
        'order_month', month(col('order_date'))
    ).withColumn(
        'order_day', dayofmonth(col('order_date'))
    )
    display(df_sales_order_header_silver)
```

```python
############################ silver layer #########################
# filter only valid sales orders use filter(col("total_due")>0)
if df_sales_order_header_silver is not None:
    df_sales_order_header_silver = df_sales_order_header_silver.filter(col("total_due") > 0)
    display(df_sales_order_header_silver)
```

```python
############################ silver layer #########################
# join customer with customer address
if df_customer_silver is not None and df_customer_address_silver is not None:
    df_customer_joined = df_customer_silver.join(
        df_customer_address_silver,
        df_customer_silver['customer_id'] == df_customer_address_silver['customer_id'],
        'inner'
    )
    display(df_customer_joined)
```

```python
############################ silver layer #########################
# join address table to get customer city and state
if df_address_silver is not None and df_customer_joined is not None:
    df_customer_joined = df_customer_joined.join(
        df_address_silver,
        df_customer_joined['address_id'] == df_address_silver['address_id'],
        'inner'
    )
    display(df_customer_joined)
```

```python
############################ silver layer #########################
#join product with product category
if df_product_silver is not None and df_product_category_silver is not None:
    df_product_joined = df_product_silver.join(
        df_product_category_silver,
        df_product_silver['product_category_id'] == df_product_category_silver['product_category_id'],
        'inner'
    )
    display(df_product_joined)
```

```python
############################ silver layer #########################
# join product_model with product_description using multi-table joins
if df_product_model_silver is not None and df_product_description_silver is not None:
    df_product_model_joined = df_product_model_silver.join(
        df_product_description_silver,
        df_product_model_silver['product_description_id'] == df_product_description_silver['product_description_id'],
        'inner'
    )
    display(df_product_model_joined)
```

```python
############################ silver layer #########################
# calculate line amount from quantity and unit price using withColumn()
if df_sales_order_detail_silver is not None:
    df_sales_order_detail_silver = df_sales_order_detail_silver.withColumn(
        'line_amount',
        col('order_qty') * col('unit_price')
    )
    display(df_sales_order_detail_silver)
```

```python
############################ silver layer #########################
# rank customer orders by order date using window function and row_number
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number

if df_sales_order_header_silver is not None:
    window_spec = Window.partitionBy('customer_id').orderBy(col('order_date').desc())
    df_sales_order_header_silver = df_sales_order_header_silver.withColumn(
        'order_rank',
        row_number().over(window_spec)
    )
    display(df_sales_order_header_silver)
```

```python
############################ silver layer #########################
# find latest customer address using window function and rank()
from pyspark.sql.window import Window
from pyspark.sql.functions import rank

if df_customer_address_silver is not None:
    window_spec = Window.partitionBy('customer_id').orderBy(col('modified_date').desc())
    df_customer_address_silver = df_customer_address_silver.withColumn(
        'address_rank',
        rank().over(window_spec)
    )
    df_latest_customer_address = df_customer_address_silver.filter(col('address_rank') == 1)
    display(df_latest_customer_address)
```

```python
############################ silver layer #########################

# Write cleaned tables as Silver Delta tables. Use write.format("delta"), mode("overwrite")
if df_address_silver is not None:
    df_address_silver.write.format("delta").mode("overwrite").saveAsTable("address_silver")
if df_customer_silver is not None:
    df_customer_silver.write.format("delta").mode("overwrite").saveAsTable("customer_silver")
if df_customer_address_silver is not None:
    df_customer_address_silver.write.format("delta").mode("overwrite").saveAsTable("customer_address_silver")
if df_product_silver is not None:
    df_product_silver.write.format("delta").mode("overwrite").saveAsTable("product_silver")
if df_product_category_silver is not None:
    df_product_category_silver.write.format("delta").mode("overwrite").saveAsTable("product_category_silver")
if df_product_description_silver is not None:
    df_product_description_silver.write.format("delta").mode("overwrite").saveAsTable("product_description_silver")
if df_product_model_silver is not None:
    df_product_model_silver.write.format("delta").mode("overwrite").saveAsTable("product_model_silver")
if df_product_model_description_silver is not None:
    df_product_model_description_silver.write.format("delta").mode("overwrite").saveAsTable("product_model_description_silver")
if df_sales_order_detail_silver is not None:
    df_sales_order_detail_silver.write.format("delta").mode("overwrite").saveAsTable("sales_order_detail_silver")
if df_sales_order_header_silver is not None:
    df_sales_order_header_silver.write.format("delta").mode("overwrite").saveAsTable("sales_order_header_silver")
```

```python
###############  Gold Layer ##################
# Create customer sales summary table. Use groupBy("customer_id").agg(sum())
from pyspark.sql.functions import sum

if df_sales_order_header_silver is not None:
    df_customer_sales_summary = df_sales_order_header_silver.groupBy("customer_id").agg(
        sum("total_due").alias("total_sales")
    )
    display(df_customer_sales_summary)
```

```python
###############  Gold Layer ##################
# Create monthly sales report. Use groupBy(year, month)
from pyspark.sql.functions import year, month, sum

if df_sales_order_header_silver is not None:
    df_monthly_sales_report = df_sales_order_header_silver.groupBy(
        year("order_date").alias("year"),
        month("order_date").alias("month")
    ).agg(
        sum("total_due").alias("monthly_sales")
    )
    display(df_monthly_sales_report)
```

```python
###############  Gold Layer ##################
## Find top 10 customers by revenue. | Use `orderBy(desc())`, `limit(10)` |
from pyspark.sql.functions import desc

if df_customer_sales_summary is not None:
    df_top_customers = df_customer_sales_summary.orderBy(desc("total_sales")).limit(10)
    display(df_top_customers)
```

```python
###############  Gold Layer ##################
## Find top 10 products by sales amount. | Use joins, `groupBy("product_name")`
from pyspark.sql.functions import sum, desc

if df_sales_order_detail_silver is not None and df_product_silver is not None:
    df_product_sales = df_sales_order_detail_silver.join(
        df_product_silver,
        df_sales_order_detail_silver['product_id'] == df_product_silver['product_id'],
        'inner'
    ).groupBy(
        df_product_silver['name'].alias('product_name')
    ).agg(
        sum('line_total').alias('total_sales')
    ).orderBy(
        desc('total_sales')
    ).limit(10)
    display(df_product_sales)
```

```python
## Create category-wise revenue report. | Join product and category tables
from pyspark.sql.functions import sum

if df_sales_order_detail_silver is not None and df_product_silver is not None and df_product_category_silver is not None:
    df_category_revenue = df_sales_order_detail_silver.join(
        df_product_silver,
        df_sales_order_detail_silver['product_id'] == df_product_silver['product_id'],
        'inner'
    ).join(
        df_product_category_silver,
        df_product_silver['product_category_id'] == df_product_category_silver['product_category_id'],
        'inner'
    ).groupBy(
        df_product_category_silver['name'].alias('category_name')
    ).agg(
        sum('line_total').alias('category_revenue')
    )
    display(df_category_revenue)
```

```python
## Calculate total orders per customer. | Use `countDistinct()`
from pyspark.sql.functions import countDistinct

if df_sales_order_header_silver is not None:
    df_customer_order_count = df_sales_order_header_silver.groupBy("customer_id").agg(
        countDistinct("sales_order_id").alias("total_orders")
    )
    display(df_customer_order_count)
```

```python
## Calculate average order value. | Use `avg("total_due")
from pyspark.sql.functions import avg

if df_sales_order_header_silver is not None:
    df_avg_order_value = df_sales_order_header_silver.groupBy("customer_id").agg(
        avg("total_due").alias("avg_order_value")
    )
    display(df_avg_order_value)
```

```python
## Find customers with no orders. | Use `left_anti` join 
from pyspark.sql.functions import countDistinct

if df_customer_silver is not None and df_sales_order_header_silver is not None:
    df_customers_with_no_orders = df_customer_silver.join(
        df_sales_order_header_silver,
        df_customer_silver['customer_id'] == df_sales_order_header_silver['customer_id'],
        'left_anti'
    ).select(
        countDistinct('customer_id').alias('customers_with_no_orders')
    )
    display(df_customers_with_no_orders)
```

```python
##Find products never sold. | Use `left_anti` join and `countDistinct()`
from pyspark.sql.functions import countDistinct

if df_product_silver is not None and df_sales_order_detail_silver is not None:
    df_products_never_sold = df_product_silver.join(
        df_sales_order_detail_silver,
        df_product_silver['product_id'] == df_sales_order_detail_silver['product_id'],
        'left_anti'
    ).select(
        countDistinct('product_id').alias('product_count')
    )
    display(df_products_never_sold)
```

```python
## Create daily sales KPI table. | Use `groupBy("order_date")`
from pyspark.sql.functions import sum, countDistinct

if df_sales_order_header_silver is not None:
    df_daily_sales = df_sales_order_header_silver.groupBy("order_date").agg(
        sum("total_due").alias("total_sales"),
        countDistinct("sales_order_id").alias("total_orders")
    )
    display(df_daily_sales)
```

```python
## Create yearly revenue trend. | Use `year()`, `sum()
from pyspark.sql.functions import year, sum

if df_sales_order_header_silver is not None:
    df_yearly_revenue = df_sales_order_header_silver.groupBy(year("order_date").alias("year")).agg(
        sum("total_due").alias("total_revenue")
    )
    display(df_yearly_revenue)
```

```python
## Calculate estimated profit with 20 percent margin. | Use `withColumn("profit", col("sales") * 0.20)`
from pyspark.sql.functions import col

if df_sales_order_header_silver is not None:
    df_profit = df_sales_order_header_silver.withColumn("profit", col("total_due") * 0.20)
    display(df_profit)
```

```python
## Create customer lifetime value table. | Use `groupBy(customer).agg(sum(total_due))
from pyspark.sql.functions import sum

if df_sales_order_header_silver is not None:
    df_customer_lifetime_value = df_sales_order_header_silver.groupBy("customer_id").agg(
        sum("total_due").alias("lifetime_value")
    )
    display(df_customer_lifetime_value)
```

```python
## Find repeat customers. | Use `groupBy().agg(countDistinct("order_id"))`
from pyspark.sql.functions import countDistinct

if df_sales_order_header_silver is not None:
    df_repeat_customers = df_sales_order_header_silver.groupBy("customer_id").agg(
        countDistinct("sales_order_id").alias("repeat_customers")
    )
    display(df_repeat_customers)
```

```python
## Create product performance dashboard table. | Revenue, quantity sold, order count
from pyspark.sql.functions import sum, countDistinct

if df_sales_order_detail_silver is not None:
    df_product_performance = df_sales_order_detail_silver.groupBy("product_id").agg(
        sum("line_total").alias("revenue"),
        sum("order_qty").alias("quantity_sold"),
        countDistinct("sales_order_id").alias("order_count")
    )
else:
    df_product_performance = spark.createDataFrame([("product_id", "revenue", "quantity_sold", "order_count")], ["product_id", "revenue", "quantity_sold", "order_count"])
display(df_product_performance)
```

```python
## Create region-wise sales ranking. | Use `Window`, `dense_rank()
from pyspark.sql.functions import dense_rank
from pyspark.sql.window import Window

if df_sales_order_header_silver is not None:
    window = Window.partitionBy("region").orderBy(col("total_sales").desc())
    df_region_sales_rank = df_sales_order_header_silver.groupBy("region").agg(
        sum("total_due").alias("total_sales")
    ).withColumn("sales_rank", dense_rank().over(window))
    display(df_region_sales_rank)
```

```python
## Find highest revenue month for each year. | Use `Window`, `rank()`
from pyspark.sql.functions import rank
from pyspark.sql.window import Window

if df_sales_order_header_silver is not None:
    window = Window.partitionBy(year("order_date")).orderBy(col("total_sales").desc())
    df_highest_revenue_month = df_sales_order_header_silver.groupBy(year("order_date").alias("year")).agg(
        sum("total_due").alias("total_sales"),
        rank().over(window).alias("rank")
    ).filter(col("rank") == 1)
    display(df_highest_revenue_month)
```

```python
## Create final star schema fact sales table. | Use fact table with customer, product, and date keys
from pyspark.sql.functions import col

if df_sales_order_header_silver is not None and df_customer_silver is not None and df_sales_order_detail_silver is not None and df_product_silver is not None:
    df_final_sales = df_sales_order_detail_silver.join(
        df_sales_order_header_silver,
        on="sales_order_id",
        how="inner"
    ).join(
        df_customer_silver,
        on="customer_id",
        how="inner"
    ).join(
        df_product_silver,
        on="product_id",
        how="inner"
    )
    display(df_final_sales)
    ## Write final star schema fact sales table to delta lake
    df_final_sales.write.format("delta").mode("overwrite").saveAsTable("sales")
```
